```python
import numpy as np
import pandas as pd
df=pd.read_csv('/content/pre_process_datasample - pre_process_datasample.xlsx - pre_process_datasample - pre_pr.csv')

df
```

```python
from sklearn.impute import SimpleImputer

features = df.iloc[:, :-1].values

age_imputer = SimpleImputer(missing_values=np.nan, strategy='mean')
salary_imputer = SimpleImputer(missing_values=np.nan, strategy='mean')

features[:, [1]] = age_imputer.fit_transform(features[:, [1]])
features[:, [2]] = salary_imputer.fit_transform(features[:, [2]])

features
```
